# ex 1
## bernoulli


```python
numsim = 1000
p = 1/3
X = np.random.random(numsim) < p

plt.hist(X.astype(int), bins=[-0.5, 0.5, 1.5], density=True) # desnity face ca sum ariilo
#plt.xticks([0,1])
plt.show()

```

## binomiala

```python
numsim = 1000

N = 10
p = 1/3
X = (np.random.random((numsim, N)) < p).sum(axis=1)

plt.hist(X.astype(int), bins = np.arange(N)-0.5, density=True)
```

## geom

```python
numsim = 10000

maxval = 0
counts = []
p = 1/5
for i in range(numsim):
  cnt = 0
  while np.random.random() > p:
    cnt += 1
  counts.append(cnt)
  if cnt > maxval:
    maxval = cnt

plt.hist(counts, bins = np.arange(maxval)-0.5, density=True)
```

## fucnctia de repartitie

1. alegem un u random in [0, 1]
2. gasim in ce subinterval se afla
3. asociez variabilei aloatoare simulate valoare


```python
numsim = 10000
p = [1/2, 1/4, 1/8, 1/16, 1/16]
pcum = np.cumsum(p)

print(pcum)

N = len(p)
print(N)

b = np.zeros(N)
for i in range(numsim):
  u = np.random.random()
  idx = 0
  while (idx<N-1 and u>=pcum[idx]):
    idx += 1
  b[idx] += 1

print(b)
plt.bar(np.arange(N), b / numsim)
plt.show()
```
