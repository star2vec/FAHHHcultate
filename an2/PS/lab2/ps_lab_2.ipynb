```python
from time import perf_counter
```

a)

```python
numsim = 10000
nrcap = 0

start = perf_counter()
frecv = []
caplist = []
for k in range(numsim):
  nrcap += np.random.random() < 0.5
  frecv.append(nrcap/(k+1))
  caplist.append(nrcap)
end = perf_counter()

plt.plot(range(numsim), frecv, color='green')
plt.show()

print('\n')
print('timp simulare: ', end-start)
timpa = end-start
```

```python
diferenta = []
for i in range(numsim):
  diferenta.append(abs(caplist[i]-i/2))

plt.plot(range(numsim), diferenta, color='green')
```
