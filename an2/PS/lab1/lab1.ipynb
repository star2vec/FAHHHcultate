```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython import display
import matplotlib.pyplot as plt

def draw_circle_and_triangle(ax):
    """
    Deseneaza un cerc de raza 1 si triunghiul echilateral inscris in el, a carui varfuri sunt definite folosind coordonate polare.
    """
    # Deseneaza cercul ca un petic (patch) aplicat axelor figurii curente
    circle = plt.Circle((0, 0), 1, color='white', fill=False)
    ax.add_patch(circle)

    # Deseneaza cele trei laturi ale triunghiului ce unesc varfurile acestuia reprezentate in coordonate polare
    plt.plot([np.cos(-np.pi/6), np.cos(np.pi/2)],
             [np.sin(-np.pi/6), np.sin(np.pi/2)], color='green')
    plt.plot([np.cos(7 * np.pi/6), np.cos(np.pi/2)],
             [np.sin(7 * np.pi/6), np.sin(np.pi/2)], color='green')
    plt.plot([np.cos(-np.pi/6), np.cos(7 * np.pi/6)],
             [np.sin(-np.pi/6), np.sin(7 * np.pi/6)], color='green')


# Initializare desen
plt.style.use('dark_background') # Dark mode pentru o vizualizare mai buna
ax = plt.gca() # Initializarea sistemului de coordonate al figurii (axes, de aici denumirea comuna de ax) cu ajutorul caruia vom construi orice grafic
ax.set_aspect('equal', 'box') # Vom reprezenta tot timpul axele de aceleasi dimensiune
ax.set_xlim((-1, 1)) # Intervalul pentru x pe axa Ox
ax.set_ylim((-1, 1)) # Intervalul pentru y pe axa Oy
draw_circle_and_triangle(ax) # Functie care ne deseneaza cercul si triunghiul din problema lui Bertrand



nrc = 1000
nr = 0
plt.ion()
for k in range(nrc):
    o1 = np.random.random() * 2 * np.pi
    o2 = np.random.random() * 2 * np.pi
    x1 = np.cos(o1)
    y1 = np.sin(o1)
    x2 = np.cos(o2)
    y2 = np.sin(o2)
    dist = np.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    if dist > np.sqrt(3):
        nr += 1
        plt.plot([x1, x2], [y1, y2], color='yellow', alpha=np.random.random()*0.5+0.1)

    else:
        plt.plot([x1, x2], [y1, y2], color='red', alpha=np.random.random()*0.5+0.1)


plt.show()
print(nr/nrc)
```

```python
num_chords = 10000
count = 0
for k in range(num_chords):
    theta = np.random.random() * 2 * np.pi
    r = np.random.random()
    x0 = r * np.cos(theta)
    y0 = r * np.sin(theta)
    phi = np.arccos(r)
    x = [np.cos(phi + theta), np.cos(-phi + theta)]
    y = [np.sin(phi + theta), np.sin(-phi + theta)]
    length = np.sqrt((x[1] - x[0])**2 + (y[1] - y[0])**2)
    if length > np.sqrt(3):
        count += 1
        if k < 1000:  # only plot the first 1000 chords
            plt.plot(x, y, color='yellow', alpha=0.1)

    else:
        if k < 1000:  # only plot the first 1000 chords
            plt.plot(x, y, color='magenta', alpha=0.1)

print("Probability = {:.4f}".format(count / (k + 1)))  # Display probability after each simulation
```
