```python
def arunca(nra, idx, l, cap):
  rn = np.random.random()
  if rn < 1/2:
    l.append('C')
  else:
    l.append('P')

  if idx>cap-1:
    nrc = 0
    for i in range(idx-cap, idx):
      if (l[i] == 'C'):
        nrc += 1
    if (nrc == cap):
      return True

  if nra==0:
    return False

  return arunca(nra-1, idx+1, l, cap)


nrt = 10000
nrsecv3 = 0
for i in range(nrt):
  if arunca(20, 0, [], 3):
    nrsecv3 += 1
print("[caz 3] ", "nrsecv3: ", nrsecv3, "ratio: ", nrsecv3/nrt)

nrsecv4 = 0
for i in range(nrt):
  if arunca(20, 0, [], 4):
    nrsecv4 += 1
print("[caz 4] ", "nrsecv4: ", nrsecv4, "ratio: ", nrsecv4/nrt)


nrt = 100000
print('\n')

nrsecv = 0
for i in range(nrt):
  l = []
  if arunca(4, 0, [], 3):
    nrsecv += 1
print("[caz 3/4] ", "nrsecv: ", nrsecv, "ratio: ", nrsecv/nrt, "corect: ", 3/16)

nrsecv = 0
for i in range(nrt):
  l = []
  if arunca(5, 0, [], 3):
    nrsecv += 1
print("[caz 3/5] ", "nrsecv: ", nrsecv, "ratio: ", nrsecv/nrt, "corect: ", 1/4)

nrsecv = 0
for i in range(nrt):
  l = []
  if arunca(6, 0, [], 3):
    nrsecv += 1
print("[caz 3/6] ", "nrsecv: ", nrsecv, "ratio: ", nrsecv/nrt, "corect: ", 1/4 + 1/16)
```
