```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython import display

import random
```

```python
nrt = 1000

rezlg = []
rez = []
for t in range(nrt):
  nrar = 300

  w0 = 1
  wnow = w0
  wlg = 0
  val = [1/2, 21/20, 3/2]
  for i in range(nrar):
    p = np.random.random()
    if p < 1/6:
      x = val[0]
    elif p < 5/6:
      x = val[1]
    else:
      x = val[2]
    wlg += np.log(x)

    wnow *= x
  rez.append(wnow)
  rezlg.append(wlg)

plt.plot(rez, alpha=0.5)
plt.plot(rezlg, alpha=0.5)
plt.show()

plt.plot(rez, alpha=0.5)
plt.plot(rezlg, alpha=0.5)
plt.ylim(min(rezlg)-1, max(rezlg)+1)
plt.show()
print((31/30) ** 300)

plt.subplot(1, 2, 1)
plt.hist(rezlg, bins=27)
plt.subplot(1, 2, 2)
plt.hist(rezlg, bins=10)
plt.show()
print(np.mean(rezlg))
```
