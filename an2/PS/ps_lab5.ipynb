# ex 1

```python
# ignoram posibilitatea falimentului
def simulare1(wealth=100, p=1/6, ppg=10): # de pus si p=1/2; ppg - price per game
  desired_sum = 10*wealth
  nrgames = 0
  while (wealth < desired_sum):
    nrgames += 1
    wealth -= ppg
    nr = 0
    while (np.random.random() > p):
        nr = nr + 1
    wealth += 2**nr

  return nrgames


# exista sansa de faliment
def simulare2(wealth=100, p=1/6, ppg=10):
  desired_sum = 10*wealth
  nrgames = 0

  while ((wealth < desired_sum) and (wealth > 0)):
    nrgames += 1
    wealth -= ppg
    nr = 0
    while (np.random.random() > p):
        nr = nr + 1
    wealth += 2**nr

  if (wealth <= 0):
    pierdut = 1
  else:
    pierdut = 0

  return nrgames, pierdut

nrt = 100000
nrp = 0
runde1 = 0
runde2 = 0
for i in range(nrt):
  rez, loss = simulare2()
  r1 = simulare1()
  runde1 += r1
  runde2 += rez
  nrp += loss
  if (i<10):
    print(r1, rez)

print()
print("medie runde: ", runde1/nrt, runde2/nrt)

print()
print(nrp, " / ", nrt)

print("probabilitate pierdere: ", nrp / nrt)
```

```python
def simbev(w=61, p=1/6):
  prizes = [1, 2, 6, 22, 200, 1000000]
  nrgames = 0

  bestscore = -1
  beststate = -1
  for ppg in [62, 100]:
    p = 1/6
    bev = 1
    e = 0
    for i in range(6):
      bev *= ((w-ppg+prizes[i]) ** p)
      e += (w-ppg+prizes[i]) / p
    e = round(e, 2)
    print("bev: ", bev, "e: ", e)
    if (bev - w < bestscore):
      bestscore = bev - w
      beststate = ppg

  return beststate

simbev()
```
