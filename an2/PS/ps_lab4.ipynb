```python
nrt = 100000

nrs = 0
nrr = 0
for i in range(nrt):
  masina = np.random.randint(3)
  usa = 1

  if masina == 1:
    nrr += 1
    goat = np.random.randint(2, 4)

    switch = 2 + 3 - goat
    if switch == masina:
      nrs += 1

  else:
    goat = 2 + 3 - masina
    switch = 2 + 3 - goat
    if switch == masina:
      nrs += 1

print("win ramanem: ", nrr, "ratio: ", nrr/nrt)
print("win switch: ", nrs, "ratio: ", nrs/nrt)
```

```python
nrt = 10000

N = 1000
K = np.random.randint(1, N-1)
print("N: ", N, "K: ", K)
nrs = 0
nrr = 0
for i in range(nrt):
  masina = np.random.randint(1, N+1)
  usa = 1

  if masina == 1:
    nrr += 1

  else:
    optiuni = []
    for i in range(2, N+1):
      if i != masina:
        optiuni.append(i)

    inchise = random.sample(optiuni, N-K-1)
    inchise.append(masina)
    switch = np.random.choice(inchise)
    if switch == masina:
      nrs += 1

print("win ramanem: ", nrr, "ratio: ", nrr/nrt)
print("win switch: ", nrs, "ratio: ", nrs/nrt)
```
